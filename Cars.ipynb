### 2. Modifying or Removing unwanted data.

```python
# We want to remove those columns which do not contribute to our analysis. 
# For this we will use drop() method.
# To remove one column, simply pass the column name in the parameter of drop method.
# To remove multiple columns, put all the columns in the list and pass it as a parameter to drop() method.
cars = cars.drop(['MSRP','Invoice'],axis = 1)
```

```python
cars
```

```python
# To keep only one row and remove all other duplicates, we can use drop_duplicates() method
cars = cars.drop_duplicates(keep = 'first') # Keeps first values and drops the rest
```

```python
cars
```

```python
# Now we can find how many rows are having missing values. 
cars.isnull().sum()
```

```python
cars.dropna(inplace=True) # inplace = True means removes from original dataset ie rows are deleted in same dataframe
```

```python
cars
```

```python
# We need to sort the rows depending on the columns
cars_sort = cars.sort_values(by = 'MPG_City',ascending = False)
```

```python
cars_sort
```

```python
# Let us know how to modify the rows in a column
# For example, there's a column : origin which contains the name of the continents where the cars are manufactured. 
# First we'll find out how many continents are there in this column by applying unique() function.
cars['Origin'].unique()
```

```python
# Since there are three continents, we can convert them into numbers like 1,2,3. 
# Convert the continent names into numeric labels 
# Asia = 1, Europe = 2, USA = 3
def origin_label(x):
    if x == 'Asia':
        return 1 
    if x == 'Europe':
        return 2
    if x == 'USA':
        return 3
```

```python
# Now apply the origin columns to these function using apply() function.
cars['Origin']=cars['Origin'].apply(origin_label)
```

```python
cars
```

### 4. Getting statistical information. 

```python
# describe() function to obtain statistical information like percentile, mean, standard deviation etc.
cars.describe()
```

```python
# To know the relationship between the numeric columns only , use the corr() method.
cars2=cars.select_dtypes(include = ['float64','int64'])
cars2.corr(method='pearson')
```

```python
# The positive number indicates positive correlation .
# Positive Correlation means If one variable increases, the other variable will also increase. 
# Negative number indicates if one variable increases , the other variable should decrease. 
# To find correlation between all columns and a particular column
cars2.corr(method='pearson')['MPG_City']
```
